```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
import plotly.express as px
print('Done')

# Replace 'your_dataset_name.csv' with the actual name of your file
# You can find the path by navigating your Drive in the Colab file browser
file_path = '/content/drive/MyDrive/Health_AnimalBites.csv'

try:
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please check the path and filename.")
except Exception as e:
    print(f"An error occurred while loading the dataset: {e}")
```

```python
df['bite_date'] = pd.to_datetime(df['bite_date'], errors='coerce')
df.dropna(subset=['bite_date'], inplace=True)
df['bite_year'] = df['bite_date'].dt.year

# Filter the DataFrame to include only years from 2010 to 2017
df_filtered_by_year = df[(df['bite_year'] >= 1985) & (df['bite_year'] <= 2017)].copy()

bites_per_year = df_filtered_by_year['bite_year'].value_counts().sort_index().reset_index()
bites_per_year.columns = ['year', 'num_bites']

fig = px.line(bites_per_year, x='year', y='num_bites',
              title='Number of Animal Bites Per Year (1985-2017)',
              labels={'year': 'Year', 'num_bites': 'Number of Bites'})
fig.update_xaxes(dtick="M1", tickformat="%Y", showgrid=True)
fig.update_yaxes(showgrid=True)
fig.show()
```

```python
species_bites = df_filtered_by_year['SpeciesIDDesc'].value_counts().reset_index()
species_bites.columns = ['Species', 'Number of Bites']

# Filter to include only Dog, Cat, and Bat
species_bites_filtered = species_bites[species_bites['Species'].isin(['DOG', 'CAT', 'BAT'])]

fig_species = px.bar(species_bites_filtered, x='Species', y='Number of Bites',
                     title='Number of Animal Bites by Species (Dog, Cat, Bat) (1985-2017)',
                     labels={'Species': 'Animal Species', 'Number of Bites': 'Count'})
fig_species.update_xaxes(tickangle=45)
fig_species.show()
```

```python
body_parts = df_filtered_by_year['WhereBittenIDDesc'].value_counts().reset_index()
body_parts.columns = ['Body Part', 'Number of Bites']

fig_pie = px.pie(body_parts, values='Number of Bites', names='Body Part',
                 title='Body Parts Most Frequently Bitten (1985-2017)')
fig_pie.show()
```
