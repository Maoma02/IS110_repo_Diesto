# **CHAPTER 4 EXERCISE**

Rewrite your pay computation with time-and-a-half for overtime and create a function called compute pay which takes two parameters (hours and rate).

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    


```python
#CORRECTED OUTPUT

def compute_pay(hours, rate):
    if hours > 40:
        # Regular pay for 40 hours
        regular_pay = 40 * rate
        # Overtime pay (time-and-a-half)
        overtime_pay = (hours - 40) * (rate * 1.5)
        total_pay = regular_pay + overtime_pay
    else:
        # No overtime
        total_pay = hours * rate
    return total_pay

# Ask user for input
hours = float(input("Enter Hours: "))
rate = float(input("Enter Rate: "))

pay = compute_pay(hours, rate)
print("Pay:", pay)
```

# **CHAPTER 4 CHALLENGE**

Write a function called computegrade that takes a score as its parameter and returns a grade as a string using the following table:

    Score           Grade
    [ 97, 100]      1.0
    [ 93,  97)		1.25
    [ 89,  93)		1.5
    [ 85,  89)		1.75
    [ 81,  85)		2.0
    [ 77,  81)		2.25
    [ 74,  77)		2.5
    [ 0,   74)		3.0

    Run the program repeatedly to test the various different values for input.
    

```python
score = float(input("Enter score: "))
print("Score:", score)

if score >= 97 and score <= 100:
    print("Grade: 1.0")
elif score >= 93 and score < 97:
    print("Grade: 1.25")
elif score >= 89 and score < 93:
    print("Grade: 1.5")
elif score >= 85 and score < 89:
    print("Grade: 1.75")
elif score >= 81 and score < 85:
    print("Grade: 2.0")
elif score >= 77 and score < 81:
    print("Grade: 2.25")
elif score >= 74 and score < 77:
    print("Grade: 2.5")
elif score >= 0 and score < 74:
    print("Grade: 3.0")
else:
    print("Invalid score")

while True:
    try:
        score = float(input("Enter score (or -1 to quit): "))
        if score == -1:
            print("Exiting program.")
            break
        grade = computegrade(score)
        print("Score:", score, "Grade:", grade)
    except ValueError:
        print("Please enter a valid number.")
```
